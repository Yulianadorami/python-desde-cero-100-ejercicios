## 7. Cuando algo falla

Te vas a encontrar con errores en rojo. Es normal y le pasa a todo el mundo. Los tres más comunes al empezar:

| Error | Qué significa | Cómo se arregla |
|---|---|---|
| `NameError` | Usaste una variable que no existe | Revisa cómo la escribiste, o ejecuta la celda que la crea |
| `IndentationError` | Problema con los espacios al inicio de la línea | Usa 4 espacios dentro de `if`, `for` y `def` |
| `SyntaxError` | Falta algo: dos puntos, comillas, un paréntesis | Mira la línea que señala el error, y también la anterior |

**Lee siempre la última línea del error:** ahí está la explicación real. Todo lo de arriba es el rastro de cómo llegó hasta ahí.
