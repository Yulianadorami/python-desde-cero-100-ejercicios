Se combinan condiciones con `&` (y) o `|` (o). **Cada condición va entre paréntesis.**


## Ejercicio 100 · Resumen estadístico completo

### 📖 Ejemplo resuelto

`.describe()` calcula de golpe conteo, promedio, desviación, mínimo, máximo y cuartiles.

```python
import pandas as pd

df = pd.DataFrame({"precio": [100, 200, 300]})
print(df["precio"].describe())
```

**Resultado:**

```
count      3.0
mean     200.0
std      100.0
min      100.0
25%      150.0
50%      200.0
75%      250.0
max      300.0
Name: precio, dtype: float64
```

> El `50%` es la mediana. Compararla con `mean` revela si hay valores atípicos.
