## Ejercicio 40 · Qué hay en común

### 📖 Ejemplo resuelto

`&` devuelve los elementos que están en ambos sets; `|` devuelve todos los de los dos.

```python

b = {2, 3, 4}
print(a & b)

```

**Resultado:**

```

{1, 2, 3, 4}
```

> `&` es intersección (lo común) y `|` es unión (todo junto, sin repetir).
